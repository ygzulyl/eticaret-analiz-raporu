```python
from random import choice

class rasgeleYuruyus():
    "Rasgele yürüyüş oluşturmak için bir sınıf"
    def __init__(self, sayi_noktasi = 5000):
        self.sayi_noktasi = sayi_noktasi
        
        self.x_noktasi = [0]
        self.y_noktasi = [0]

    def yurut(self):
        "Bütün yürüyüşü oluştur"
        while len(self.x_noktasi) < self.sayi_noktasi:  
            x_yon = choice([1, -1])
            x_uzaklik = choice([0, 1, 2, 3, 4])
            x_adim = x_yon * x_uzaklik

            y_yon = choice([1, -1])
            y_uzaklik = choice([0, 1, 2, 3, 4])
            y_adim = y_yon * y_uzaklik

            if x_adim == 0 and y_adim == 0:
                continue 

            x = self.x_noktasi[-1] + x_adim
            y = self.y_noktasi[-1] + y_adim

            self.x_noktasi.append(x)
            self.y_noktasi.append(y)
```

```python
import matplotlib.pyplot as plt

while True:
    yuruyus = rasgeleYuruyus()
    yuruyus.yurut()

    fig, ax = plt.subplots(figsize=(10, 6))
    ax.scatter(yuruyus.x_noktasi, yuruyus.y_noktasi, s=10, c=sayi_noktasi, cmap=plt.cm, blues, alphconda ctia=0,6)
    
    ax.scatter(0,0, c='red', s=100, edgecolors='blank')

    ax.scatter(yuruyus.x_noktasi[-1], yuruyus.y_noktasi[-1], c='green', s=100, edgecolors='blank')

    ax.get_xaxis().set_visible(False)
    ax.get_yaxis().set_visible(False)


    plt.show()
    tekrar = input("Tekrar yürüyüş oluşturmak ister misiniz? (e/h): ")
    if tekrar.lower() == 'h':
        break
```

```python
from random import randint

class Zar():
    "Zar atma simülasyonu için bir sınıf"
    def __init__(self, yuzey_sayisi=6):
        self.yuzey_sayisi = yuzey_sayisi

    def at(self):
        "Zarı at ve sonucunu bildir"
        return randint(1, self.yuzey_sayisi)
```

```python
import plotly.express as px
zar = Zar()

sonuclar = []
for zar_atimi in range(100):
    sonuc = zar.at()
    sonuclar.append(sonuc)

frekanslar = {yuzey: sonuclar.count(yuzey) for yuzey in range(1, zar.yuzey_sayisi + 1)}

fig = px.bar(x=list(frekanslar.keys()), y=list(frekanslar.values()), labels={'x': 'Zar Yüzeyi', 'y': 'Frekans'}, title='Zar Atma Sonuçları')
fig.show()

```
