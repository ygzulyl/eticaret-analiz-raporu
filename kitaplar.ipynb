```python
kitaplar = ["Nutuk", "Simyacı", "Suç ve Ceza"] 

print("--- KÜTÜPHANE YÖNETİM SİSTEMİ ---")

devam_etsin_mi = True
while devam_etsin_mi:
    print("\n1-Listele | 2-Ekle | 3-Çıkar | 4-Çıkış")
    secim = input("Yapmak istediğiniz işlemi seçin: ")

    if secim == "1":
        print("\nGüncel Kitap Listesi:")
        for kitap in kitaplar:
            print(f"- {kitap.title()}") 
            
    elif secim == "2":
        yeni_kitap = input("Eklemek istediğiniz kitabın adı: ")
        
        if yeni_kitap.title() in [k.title() for k in kitaplar]:
            print(f"HATA: '{yeni_kitap.title()}' zaten mevcut!")
        else:
            kitaplar.append(yeni_kitap)
            print(f"'{yeni_kitap.upper()}' başarıyla eklendi.")

    elif secim == "3":
        silinecek = input("Çıkarmak istediğiniz kitabın tam adını yazın: ")
        
        bulundu = False
        for kitap in kitaplar:
            if kitap.lower() == silinecek.lower():
                kitaplar.remove(kitap)
                print(f"'{kitap.title()}' listeden çıkarıldı.")
                bulundu = True
                break 
        
        if not bulundu:
            print("HATA: Aradığınız kitap kütüphanede bulunamadı.")

    elif secim == "4":
        print("Programdan çıkılıyor... İyi okumalar!")
        devam_etsin_mi = False
        
    else:
        print("Geçersiz seçenek, tekrar deneyin.")
```
