```python
sayi = int(input("Lütfen bir sayı girin: "))
ciftler = 0
tekler = 0

for i in range(1, sayi):
    if i % 2 == 0:
        ciftler += i
    else:
        tekler += i

print("Çift sayıların toplamı:", ciftler)
print("Tek sayıların toplamı:", tekler)
```

```python
onaylanmamis_kullanicilar = ['alice', 'brian', 'candace']
onaylanmis_kullanicilar = []

while onaylanmamis_kullanicilar:
    mevcut_kullanici = onaylanmamis_kullanicilar.pop()
    print("Onaylanıyor:", mevcut_kullanici)
    onaylanmis_kullanicilar.append(mevcut_kullanici)

print("\nOnaylanmış kullanıcılar:", onaylanmis_kullanicilar)
print("Onaylanmamış kullanıcılar", onaylanmamis_kullanicilar)
```

```python
yanitlar = {}

anket_aktif = True

while anket_aktif:
    name = input("\nAdınız nedir? ")
    yanit = input("Günün birinde hangi dağa çıkmak istersiniz? ")
    yanitlar[name] = yanit

    devam = input("Başka biri katılacak mı? (E/H) ")
    if devam.lower() == 'h':
        anket_aktif = False

print("\n--- Anket Sonuçları ---")

for name, yanit in yanitlar.items():
    print(f"{name} {yanit} dağına çıkmak istiyor: ")
```
