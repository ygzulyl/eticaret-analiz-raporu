```python
from random import randint

class Zar:
    """Bir zar için bir sınıf"""
    def __init__(self, zar_yuzu = 6):
        """Zarın varsayılan değerini 6 yapar."""
        self.zar_yuzu = zar_yuzu

    def zar_at(self):
        """1 ile zarın yüz sayısı arasında rastgele bir değer döndürür."""
        return randint(1, self.zar_yuzu)
```

```python
import plotly.express as px
zar = Zar()

sonuclar = []

for zar_atma in range(10000):
    sonuc = zar.zar_at()
    sonuclar.append(sonuc)

frekanslar = []
for zar_yuzu in range(1, zar.zar_yuzu + 1):
    frekans = sonuclar.count(zar_yuzu)
    frekanslar.append(frekans)

x_degerleri = list(range(1, zar.zar_yuzu + 1))
data = (x_degerleri, frekanslar)

fig = px.bar(x=x_degerleri, y=frekanslar, labels={'x': 'Zar Yüzü', 'y': 'Frekans'}, title= 'Zar Atma Sonuçları')
fig.show()
```

```python
zar1 = Zar(zar_yuzu=20)
zar2 = Zar(zar_yuzu=20)

sonuclar = []

for zar_atma in range(100000):
    sonuc1 = zar1.zar_at()
    sonuc2 = zar2.zar_at()
    toplam_sonuc = sonuc1 + sonuc2
    sonuclar.append(toplam_sonuc)

frekanslar = []
for toplam in range(2, 41):
    frekans = sonuclar.count(toplam)
    frekanslar.append(frekans)

x_degerleri = list(range(2, 41))
data = [x_degerleri, frekanslar]

fig = px.bar(x=x_degerleri, y=frekanslar, labels={'x': 'Toplam', 'y': 'Frekans'}, title='Zar Atma Sonuçları')
fig.show()
```

```python
def fahrenheit_to_celcius(fahrenheit):
    return (fahrenheit - 32) * 5/9
```

```python
import csv
import matplotlib.pyplot as plt
import datetime 

dosya_adi = "sitka_weather_2021_simple.csv"
with open(dosya_adi) as dosya:
    csv_okuyucu = csv.reader(dosya)
    basliklar = next(csv_okuyucu)
    print(basliklar)

    yuksek_sicakliklar = []
    tarihler = []
    dusuk_sicakliklar = []
    for satir in csv_okuyucu:
        tarih = datetime.datetime.strptime(satir[2], "%Y-%m-%d")
        tarihler.append(tarih)
        yuksek_sicaklik = fahrenheit_to_celcius(int(satir[4]))
        yuksek_sicakliklar.append(yuksek_sicaklik)
        dusuk_sicaklik = fahrenheit_to_celcius(int(satir[5]))
        dusuk_sicakliklar.append(dusuk_sicaklik)
        
plt.style.use('seaborn-v0_8')
fig, ax = plt.subplots()
ax.plot(tarihler, yuksek_sicakliklar, c='red')
ax.plot(tarihler, dusuk_sicakliklar, c='blue')
plt.title("Günlük Yüksek ve Düşük Sıcaklıklar, 2021, fontsize=24")
plt.xlabel(' ', fontsize=16)
fig.autofmt_xdate()
plt.ylabel("Sıcaklık (F), fontsize=16")
plt.tick_params(axis='both', which='major', labelsize=16)
plt.show()
```
