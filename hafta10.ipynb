```python
yas = int(input("Lütfen yaşınızı giriniz."))

if yas < 0 or yas > 120:
    print("Geçersiz yaş girdiniz.")
else:
    if yas < 6:
        print("Çocuk bileti: Ücretsiz")
    elif 6 <= yas < 18:
        print("Öğrenci bileti: 20TL")
    elif 18 <= yas < 60:
        print("Yetişkin bileti: 50TL")
    elif 60 <= yas < 65:
        print("Yaşlı bileti: 30TL")
    else:
        print("Emekli bileti: Ücretsiz")
```

```python
istenilen_malzemeler = ["ananas"]
stoktaki_malzemeler = ["peynir", "sucuk", "mantar", "zeytin", "sosis"]

if not istenilen_malzemeler:
    for malzeme in istenilen_malzemeler:
        if malzeme not in stoktaki_malzemeler:
            print(f"{malzeme.title()} eklenmiyor, stoklarımızda bulunmamaktadır.")
        else:
            print("Pizzanıza (malzeme) ekleniyor.")
else:
    print("Herhangi bir malzeme seçilmedi, sade pizza hazırlanıyor.")

print("Pizzanız hazırdır!")
```
