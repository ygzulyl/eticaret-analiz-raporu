```python
uzayli_0 = {"renk": "yeşil", "puan": 5, "hız": "yavaş", "x_koordinat": 0, "y_koordinat": 0}
print(f"Orijinal konum: {uzayli_0['x_koordinat']}, {uzayli_0['y_koordinat']}")

if uzayli_0["hız"] == "yavaş":
    x_hareketi = 1
elif uzayli_0["hız"] == "orta":
    x_hareketi = 2
else:
    x_hareketi = 3

uzayli_0["x_koordinat"] += x_hareketi
print(f"Yeni konum: {uzayli_0['x_koordinat']}, {uzayli_0['y_koordinat']}")
```

```python
favori_diller = {
    "ali": "python",
    "ayşe": "java",
    "mehmet": "c++",
    "fatma": "ruby",
    "can": "python",
    "deniz": "java"}

for isim in favori_diller.keys():
    print(f"Merhabalar {isim.title()}!")
    if isim == "fatma":
        dil = favori_diller[isim]
        print(f"{isim.title()}, {dil.title()} harika bir dil!")
```

```python
for isim in favori_diller.keys():
    print(f"Merhabalar {isim.title()}!")
    dil = favori_diller[isim]
    if dil == "python":
        print(f"{isim.title()}, {dil.title()} harika bir dil!")

print("\n")

for isim, dil in favori_diller.items():
    print(f"Merhabalar {isim.title()}!")
    if dil == "python":
        print(f"{isim.title()}, {dil.title()} harika bir dil!")
```
