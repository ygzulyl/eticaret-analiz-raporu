```python
bonservis_listesi = [15, 10, 75, 30, 140]
yeni_bonservisler = []

for bonservis_ in bonservis_listesi:
    bonservis_ = bonservis_+10
    yeni_bonservisler.append(bonservis_)
print(yeni_bonservisler)
```

```python
bonservis_listesi = [["Osimhen", "Singo", "İsak"],[75, 30, 140]]
yeni_bonservisler = []

for bonservis_ in bonservis_listesi[1]:
    bonservis_ = bonservis_ +25
    print("Oyuncunun yeni bonservisi:", bonservis_)
    yeni_bonservisler.append(bonservis_)
    bonservis_listesi[1] = yeni_bonservisler

print(bonservis_listesi)

```

```python
sayilar = []
for sayi in range(20):
    sayilar.append(sayi*3)
print(sorted(sayilar, reverse=True))
```

```python
sayi = input("Bir sayı girin: ")
sayilar = [1]

for i in range(1, int(sayi)+1):
    sayilar.append(i * sayilar[i-1])
print(f"{sayi}'nın faktöriyeli = {sayilar[int(sayi)]}")
```
