```python
import os

class Kitap:
    """Kitap nesnelerini oluşturmak için kullanılan sınıf."""
    def __init__(self, ad, yazar, sayfa):
        self.ad = ad
        self.yazar = yazar
        self.sayfa = sayfa

    def formatla(self):
        return f"{self.ad}|{self.yazar}|{self.sayfa}\n"

class Kutuphane:
    def __init__(self, dosya_adi="kitaplar.txt"):
        self.dosya_adi = dosya_adi

    def kitap_ekle(self, kitap):
        """Dosya yazma (Append modu)"""
        try:
            with open(self.dosya_adi, "a", encoding="utf-8") as f:
                f.write(kitap.formatla())
            print(f"'{kitap.ad}' başarıyla dosyaya eklendi!")
        except Exception as e:
            print(f"Yazma hatası: {e}")

    def listele(self):
        """Dosya okuma ve For döngüsü kullanımı"""
        if not os.path.exists(self.dosya_adi):
            print("Kütüphane dosyası henüz oluşturulmamış (Kitap yok).")
            return

        print("\n--- Kitap Listesi ---")
        try:
            with open(self.dosya_adi, "r", encoding="utf-8") as f:
                satirlar = f.readlines()
                
                if not satirlar:
                    print("Kütüphane şu an boş.")
                    return

                sayac = 1
                for satir in satirlar:
                    veri = satir.strip().split("|")
                    if len(veri) == 3:
                        ad, yazar, sayfa = veri
                        print(f"{sayac}. {ad} - {yazar} ({sayfa} sayfa)")
                        sayac += 1
        except Exception as e:
            print(f"Okuma hatası: {e}")

def main():
    kyp = Kutuphane()
    
    while True:
        print("\n1. Kitap Ekle")
        print("2. Kitapları Listele")
        print("3. Çıkış")
        secim = input("Seçiminiz: ")

        if secim == "1":
            try:
                ad = input("Kitap Adı: ")
                yazar = input("Yazar: ")
                sayfa = int(input("Sayfa Sayısı: ")) 
                
                yeni_kitap = Kitap(ad, yazar, sayfa)
                kyp.kitap_ekle(yeni_kitap)
            except ValueError:
                print("Hata: Sayfa sayısı rakam olmalıdır!")
        
        elif secim == "2":
            kyp.listele()
        
        elif secim == "3":
            print("Programdan çıkılıyor...")
            break
        else:
            print("Geçersiz seçim, tekrar deneyin.")

if __name__ == "__main__":
    main()
```
