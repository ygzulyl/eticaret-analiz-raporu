```python
class Kopek:
    def __init__(self, isim, yas):
        self.isim = isim
        self.yas = yas
    
    def otur(self):
        print(f"{self.isim} oturuyor.")

    def havla(self):
        print(f"{self.isim} havlıyor.")
```

```python
benim_kopegim = Kopek("Karabas", 5)
benim_kopegim.otur()

arkadasimin_kopegi = Kopek("Beyaz", 3)
arkadasimin_kopegi.otur()
arkadasimin_kopegi.havla()
```

```python
class Futbolcu:
    def __init__(self, isim, yas):
        self.isim = isim
        self.yas = yas

    def ısın(self):
        print(f"{self.isim} ısınıyor.")

    def koş(self):
        print(f"{self.isim} koşuyor.")
```

```python
benim_futbolcum = Futbolcu("İlkay", 35)
benim_futbolcum.ısın()

senin_futbolcun = Futbolcu("İsmail", 26)
senin_futbolcun.koş()
```

```python
class Araba:
    def __init__(self, marka, model, yil):
        self.marka = marka
        self.model = model
        self.yil = yil
        self.km = 0

    def bilgileri_goster(self):
        uzun_format = f"{self.marka} {self.model} {self.yil} ({self.km} km)"
        return uzun_format
    
    def km_guncelle(self, km):
        if km >= self.km:
            self.km = km
        else:
            print("Hata: Kilometre geri alınamaz.")

    def km_ekle(self, km):
        if km > 0:
            self.km += km
        else:
            print("Hata: Eklenen kilometre pozitif olmalıdır.")
```

```python
ilk_arabam = Araba("Toyota", "Corolla", 2020)
ilk_arabam_bilgileri = ilk_arabam.bilgileri_goster()
print(ilk_arabam_bilgileri)
ilk_arabam.km = 15000
ilk_arabam.km_guncelle(0)
ilk_arabam.km_ekle(-500)
print(ilk_arabam.bilgileri_goster())
```

```python
class ElektrikliAraba(Araba):
    def __init__(self, marka, model, yil, batarya_kapasitesi):
        super().__init__(marka, model, yil)
        self.batarya_kapasitesi = batarya_kapasitesi
```

```python
ilk_elektriki_arabam = ElektrikliAraba("Tesla", "Model 3", "2021", 75)
print(ilk_elektriki_arabam.bilgileri_goster())

print(f"Batarya Kapasitesi: {ilk_elektriki_arabam.batarya_kapasitesi} kWh")
print(f"Batarya Kapasitesi: {ilk_arabam.batarya_kapasitesi} kWh")
```
