```python
class Araba:
    def __init__(self, marka, model, yil):
        # Arkadaşının kodundaki isim, yas, sehir yapısıyla aynı
        self.marka = marka
        self.model = model
        self.yil = yil
        self.kilometre = 0  # takipci değişkeni yerine kilometre

    def bilgileri_goster(self):
        uzun_format = f"{self.marka} {self.model} ({self.yil}) model araç {self.kilometre} km'de."
        return uzun_format

    def kilometre_guncelle(self, yeni_km):
        # takipci_guncelle mantığıyla aynı: KM geri sarılamaz
        if yeni_km >= self.kilometre:
            self.kilometre = yeni_km
        else:
            print("Hata: Kilometre bilgisi düşürülemez.")

    def km_ekle(self, eklenen_km):
        # takipci_ekle mantığıyla aynı: Sürüş mesafesi ekler
        if eklenen_km > 0:
            self.kilometre += eklenen_km
        else:
            print("Hata: Eklenen kilometre pozitif olmalıdır.")

# İlk nesne oluşturma
ilk_araba = Araba("Toyota", "Corolla", 2022)
print(ilk_araba.bilgileri_goster())

# Değerleri güncelleme
ilk_araba.kilometre = 1000
ilk_araba.kilometre_guncelle(1500)
ilk_araba.km_ekle(200)

print(ilk_araba.bilgileri_goster())

---

# Kalıtım (Inheritance) yapısı
class SporAraba(Araba):
    def __init__(self, marka, model, yil, mod):
        # super() kullanımıyla ana sınıftan özellikleri alıyoruz
        super().__init__(marka, model, yil)
        self.mod = mod  # platform değişkeni yerine sürüş modu

# Alt sınıftan nesne oluşturma
ilk_spor_araba = SporAraba("Ferrari", "F8", 2023, "Yarış Modu")

print(ilk_spor_araba.bilgileri_goster())
print(f"Sürüş Modu: {ilk_spor_araba.mod}")
```
