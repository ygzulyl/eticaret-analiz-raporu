```python
import csv
import matplotlib.pyplot as plt
import datetime

dosya_adi = "gold-silver (1).csv"
with open(dosya_adi) as dosya:
    csv_okuyucu = csv.reader(dosya)
    basliklar = next(csv_okuyucu)
    print(basliklar)

    altin_fiyatlari = []
    tarihler = []
    gumus_fiyatlari = []
    for satir in csv_okuyucu:
        tarih = datetime.datetime.strptime(satir[0], "%Y-%m-%d")
        tarihler.append(tarih)
        altin_fiyati = float(satir[1])
        altin_fiyatlari.append(altin_fiyati)
        gumus_fiyati = float(satir[2])
        gumus_fiyatlari.append(gumus_fiyati)

plt.style.use("ggplot")
fig, ax = plt.subplots()
ax.plot(tarihler, altin_fiyatlari, c="yellow")
ax.plot(tarihler, gumus_fiyatlari, c="gray")
plt.title("Altın ve Gümüş Fiyatları, 2006-2026, fontsize=14")
plt.xlabel(' ', fontsize=12)
fig.autofmt_xdate()
plt.ylabel("Fiyat (USD)", fontsize=12)
plt.tick_params(axis='both', which='major', labelsize=10)
plt.show()
```
