```python
ogrenci_isimleri_numaralari = {
    "Yusuf" : "1027",
    "Nisa" : "752",
    "Yavuz" : "861",
    "Aleyna" : "955",
    "Yağız" : "902"
}
```

```python
for ogrenci_isimleri in ogrenci_isimleri_numaralari.keys():
    print(f"öğrencinin ismi: {ogrenci_isimleri}")

for ogrenci_numaraları in ogrenci_isimleri_numaralari.values():
    print(f"öğrencinin numarası: {ogrenci_numaraları}")
```

```python
for isim, numara in ogrenci_isimleri_numaralari.items():
    print(f"{isim} isimli öğrencinin numarası: {numara.title()} 'dır.")

    if isim == "Yusuf":
        print(f"{isim.title()} ({numara}) devamsızlığı bulunmaktadır.")
```
