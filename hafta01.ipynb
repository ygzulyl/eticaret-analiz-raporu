```python
def profil_olustur(ad:str, soyad:str, **kullanici_bilgileri):
    """Kullanıcı bilgilerini içeren bir profil oluşturur."""
    profil = {}
    profil['ad'] = ad
    profil['soyad'] = soyad
    for anahtar, deger in kullanici_bilgileri.items():
        profil[anahtar] = deger

    return profil
```

```python
kubilayerislikprofil = profil_olustur("Kubilay", "Erislik", yas=30, meslek="Mühendis")
print(kubilayerislikprofil)
```
