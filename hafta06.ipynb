```python
while True:
    try:
        birinci_sayi = int(input("Birinci sayıyı giriniz: "))
        if birinci_sayi.lower() == "q":
            print("Çıkış yapılıyor...")
        ikinci_sayi = int(input("İkinci sayıyı giriniz: "))
        sonuc = birinci_sayi / ikinci_sayi
        print(f"Gerçekleştirilen bölme işleminin sonucu: {sonuc}")
    except ValueError:
        print("Lütfen geçerli bir sayı giriniz.")
    except ZeroDivisionError:
        print("Sıfıra bölme hatası: İkinci sayı sıfır olamaz.")
    except:
        print("Beklenmeyen bir hata oluştu.")

print("Program sonlandı")
```

```python
dosya_yolu = input("Okunacak dosyanın adını giriniz.")

try:
    with open(dosya_yolu, encoding="UTF-8") as dosya:
        icerik = dosya.read()
except FileNotFoundError:
    print(f"{dosya_yolu} dosyası bulunamadı.")
else:
    kelimeler = icerik.split()
    kelime_sayisi = len(kelimeler)
    print(f"{dosya_yolu} dosyasında toplam {kelime_sayisi} kelime bulunmaktadır.")
    harcanacak_sure = kelime_sayisi / 200
    if harcanacak_sure < 60:
        print(f"Bu dosyayı okumak yaklaşık {harcanacak_sure: .2f} dakika sürecektir.")
    else:
        harcanacak_sure = harcanacak_sure / 60
        print(f"Bu dosyayı okumak yaklaşık {harcanacak_sure: .2f} dakika sürecektir.")
```

```python
def sozcuk_say(dosya_yolu):

    try: 
        with open(dosya_yolu, encoding="UTF-8") as dosya:
            icerik = dosya.read()
    except FileNotFoundError:
        print(f"{dosya_yolu} dosyası bulunamadı.")
    else:
        kelimeler = icerik.split()
        kelime_sayisi = len(kelimeler)
        print(f"{dosya_yolu} dosyasında toplam {kelime_sayisi} kelime bulunmaktadır.")
        harcanacak_sure = kelime_sayisi / 200
        if harcanacak_sure < 60:
            print(f"{dosya_yolu.split('.')[0]} kitabını okumak yaklaşık {harcanacak_sure: .2f} dakika sürecektir.")
        else:
            harcanacak_sure = harcanacak_sure / 60
            print(f"{dosya_yolu.split('.')[0]} okumak yaklaşık {harcanacak_sure: .2f} dakika sürecektir.")
```

```python
dosyalar = ["alice.txt", "arabam.txt", "moby_dick.txt", "little_women.txt"]

for dosya in dosyalar:
    sozcuk_say(dosya)
```
