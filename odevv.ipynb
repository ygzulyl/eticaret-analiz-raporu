```python
class Telefon:
    def __init__(self, marka, model, yil):
        self.marka = marka
        self.model = model
        self.yil = yil
        self.pil = 0

    def bilgileri_goster(self):
        uzun_format = f"{self.marka} {self.model} {self.yil} ({self.pil} pil)"
        return uzun_format
    
    def pil_guncelle(self, pil):
        if pil <= self.pil:
            self.pil = pil
        else:
            print("Hata: Pil geri alınamaz.")

    def pil_ekle(self, pil):
        if pil > 0:
            self.pil += pil
        else:
            print("Hata: Eklenen pil pozitif olmalıdır.")
```

```python
ilk_telefonum = Telefon("Redmi", "Note 10", 2021)
ilk_telefonum_bilgileri = ilk_telefonum.bilgileri_goster()
print(ilk_telefonum_bilgileri)
ilk_telefonum.pil = 100
ilk_telefonum.pil_guncelle(15)
ilk_telefonum.pil_ekle(50)
print(ilk_telefonum.bilgileri_goster())
```

```python
class İosTelefon(Telefon):
    def __init__(self, marka, model, yil, pil_kapasitesi):
        super().__init__(marka, model, yil)
        self.pil_kapasitesi = pil_kapasitesi
```

```python
ilk_ios_telefonum = İosTelefon("Iphone", "16 Pro Max", "2024", 100)
print(ilk_ios_telefonum.bilgileri_goster())

print(f"Pil Kapasitesi: {ilk_ios_telefonum.pil_kapasitesi} pil")
print(f"Pil Kapasitesi: {ilk_telefonum.pil_kapasitesi} pil")
```
