```python
vize_notlari = [70, 85, 90, 60, 75, 80]
yeni_notlar = []

for not_ in vize_notlari:
    not_ = not_-5
    yeni_notlar.append(not_)
print(yeni_notlar)
```

```python
not_listesi = [["Ahmet", "Mehmet", "Ayşe"],[70, 85, 90]]
yeni_notlar = []

for not_ in not_listesi[1]:
    not_ = not_ + 5
    print("Öğrencinin yeni notu:", not_)
    yeni_notlar.append(not_)
    not_listesi[1] = yeni_notlar

print(not_listesi)
```

```python
sayilar = []
for sayi in range(10):
    sayilar.append(sayi*2)
print(sorted(sayilar, reverse=True))
```

```python
sayi = input("Bir sayı girin: ")
sayilar = [1]

for i in range(1, int(sayi)+1):
    sayilar.append(i * sayilar[i-1])
print(f"{sayi}'nın faktöriyeli = {sayilar[int(sayi)]}")
```
